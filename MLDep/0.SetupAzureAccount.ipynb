#step-by-step guide to **create a free Azure account** and **set it up** properly:

---

##  Step 1: Visit Azure Free Account Page

Go to: [https://azure.microsoft.com/en-us/free](https://azure.microsoft.com/en-us/free)

---

##  Step 2: Click "Start for Free"

You’ll be redirected to the Microsoft sign-in page.

---

##  Step 3: Sign In or Create a Microsoft Account

* If you already have a Microsoft account (Outlook, Hotmail, Xbox, etc.), sign in.
* If not, click **“Create one!”** and register using:

  * Your email address (can be Gmail or any provider)
  * A strong password

---

##  Step 4: Verify Identity

To prevent fraud, Microsoft requires:

###  Phone Verification

* Enter your mobile number
* Choose **text** or **call** option to receive a verification code

###  Credit/Debit Card Verification *(Mandatory, but won’t be charged)*

* Enter your card details
* ₹2 (or equivalent) is deducted temporarily and refunded

 **No prepaid cards** are accepted (only credit/debit cards that support international transactions).

---

##  Step 5: Accept Terms & Start

* Agree to the **subscription agreement**, **offer details**, and **privacy statement**
* Click **"Sign Up"**

---

##  Done! Welcome to Azure

You now have:

| Resource                       | Limit                                            |
| ------------------------------ | ------------------------------------------------ |
| ₹15,500 (approx. \$200) credit | Valid for 30 days                                |
| 12 months of free services     | Includes VMs, storage, DB                        |
| Always-free tier               | Limited services like Azure Functions, Cosmos DB |

---

##  Step 6: Initial Setup Checklist

### 1. Go to Azure Portal

 [https://portal.azure.com](https://portal.azure.com)

### 2. Set up Resource Groups

* Navigate to **Resource Groups**
* Create one with:

  * Name: e.g., `my-first-group`
  * Region: Close to your users (e.g., Central India)

### 3. Deploy a Simple Service (Optional)

* Try creating a **Virtual Machine**, **Storage Account**, or **Web App**
* Azure has guided **Quickstarts**

### 4. Set Spending Alerts

* Go to **Cost Management → Budgets**
* Create a **budget** (e.g., ₹1000/month)
* Set **email alerts** to avoid exceeding limits
