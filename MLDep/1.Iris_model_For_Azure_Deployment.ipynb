```python


from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
import pickle


```

```python
# 1️ Load the iris dataset (150 samples, 4 features each)
iris = load_iris()
X, y = iris.data, iris.target

```

```python
print(" Input Features (X):")
print(X[:5])  # Show first 5 rows
print("Shape of X:", X.shape)  # (150, 4)
```

```python
print("\n Target Labels (y):")
print(y[:10])  # Show first 10 labels
print("Shape of y:", y.shape)  # (150,)

```

```python
# 2️ Train a Random Forest Classifier
clf = RandomForestClassifier(n_estimators=100)
clf.fit(X, y)

print("\n Model training complete.")

```

```python
# 3️ Test model on a sample input
sample = [[5.1, 3.5, 1.4, 0.2]]  # Example input
prediction = clf.predict(sample)

print("\n Sample Input for Prediction:")
print(sample)
print("Prediction Output:", prediction)  # Output: [0] (means class 0 = setosa)

```

```python
# 4️ Save the trained model using pickle
import os
os.makedirs("model", exist_ok=True)
```

```python

with open('model/iris_model.pkl', 'wb') as f:
    pickle.dump(clf, f)

print("\n Model saved to 'model/iris_model.pkl'")
```
